## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Create a copy of the original dataset
df_clean = df.copy()

# Remove the index-like column
df_clean.drop(columns=['Unnamed: 0'], inplace=True)

# Remove leading/trailing spaces from column names
df_clean.columns = df_clean.columns.str.strip()

# Remove leading/trailing spaces from text columns
text_columns = df_clean.select_dtypes(include='object').columns

for column in text_columns:
    df_clean[column] = df_clean[column].str.strip()

# Check the cleaned dataset
print("Dataset shape:", df_clean.shape)
print("\nColumn names:")
print(df_clean.columns.tolist())
````

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Based on our visualization experiments, we will test these three statements:

| Hypothesis       | Research question                                                              | Statistical test         |
| ---------------- | ------------------------------------------------------------------------------ | ------------------------ |

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Research question

Does company rating have a significant linear relationship with average salary?

Null Hypothesis (H₀):

There is no statistically significant linear relationship between Company Rating and Average Salary in the dataset.

Alternative Hypothesis (H₁):

There is a statistically significant linear relationship between Company Rating and Average Salary in the dataset.

We will use a significance level of:

`α = 0.05`

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import pearsonr
````

````python
# Perform Statistical Test to obtain P-Value

# Hypothesis 1: Pearson Correlation Test

# Select the two variables
rating_salary = df_clean[['Rating', 'Average Salary']].dropna()

# Perform Pearson correlation test
correlation, p_value = pearsonr(
    rating_salary['Rating'],
    rating_salary['Average Salary']
)

print("Pearson Correlation:", round(correlation, 4))
print("P-Value:", round(p_value, 4))

# Decision based on significance level
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis (H₀).")
    print("There is a statistically significant relationship between Rating and Average Salary.")
else:
    print("Fail to Reject the Null Hypothesis (H₀).")
    print("There is not enough statistical evidence of a significant relationship between Rating and Average Salary.")
````

##### Which statistical test have you done to obtain P-Value?

Pearson correlation test was used to obtain the p-value.

##### Why did you choose the specific statistical test?

Pearson correlation test was chosen because both Company Rating and Average Salary are numerical variables. The purpose of the test is to determine whether there is a statistically significant linear relationship between company rating and average salary.

The test provides a correlation coefficient to show the strength and direction of the relationship and a p-value to determine whether the observed relationship is statistically significant.

**Final conclusion**


The Pearson correlation coefficient is 0.0061, which is very close to zero, indicating an extremely weak linear relationship between Company Rating and Average Salary. The p-value is 0.8697, which is greater than the significance level of 0.05. Therefore, we fail to reject the null hypothesis.

In simple terms, the analysis does not provide sufficient statistical evidence that a higher company rating is associated with a higher salary in this dataset.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Research Question

Does average salary differ significantly across different company-size categories?



Null Hypothesis (H₀):

The average salaries are the same across the different company-size categories.

Alternative Hypothesis (H₁):

At least one company-size category has a different average salary.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

from scipy.stats import f_oneway

# Remove placeholder company-size categories
valid_size_data = df_clean[
    ~df_clean['Size'].isin(['-1', 'Unknown'])
].copy()

# Create salary groups for each company-size category
salary_groups = [
    group['Average Salary'].dropna()
    for _, group in valid_size_data.groupby('Size')
]

# Perform One-Way ANOVA
f_statistic, p_value = f_oneway(*salary_groups)

print("F-Statistic:", round(f_statistic, 4))
print("P-Value:", round(p_value, 4))
````

##### Which statistical test have you done to obtain P-Value?

One-Way ANOVA (Analysis of Variance) was used because Company Size is a categorical variable with multiple groups, while Average Salary is a numerical variable. The test helps determine whether there is a statistically significant difference in average salary across the different company-size categories.

##### Why did you choose the specific statistical test?

One-Way ANOVA was chosen because Company Size is a categorical variable with multiple groups, while Average Salary is a numerical variable. The test is appropriate for determining whether there is a statistically significant difference in average salary across these different company-size groups.

**Final Conclusion**

Since:

`P-value < 0.05`

we Reject the Null Hypothesis (H₀).

There is statistically significant evidence that average salary differs across company-size categories in this dataset.
